```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
vg=pd.read_csv('https://raw.githubusercontent.com/SinhaPushkar047/ML_Topicwise_Practice/refs/heads/main/Univariate_Analysis/video_game_sales_eda_practice.csv').drop(columns='game_name')
```

```python
cat_col=[col for col in vg.columns if vg[col].dtype=='object']
```

```python
num_col=[col for col in vg.columns if vg[col].dtype=='float64']
```

```python
for i in cat_col:
  sns.countplot(vg[i],palette='viridis')
  plt.show()
```

```python
for i in cat_col:
  vg[i].value_counts().plot(kind='pie',autopct='%.2f')
  plt.show()
```

```python
for i in num_col:
  sns.distplot(vg[i])
  plt.show()
```

```python
for i in num_col:
  sns.histplot(vg[i])
  plt.show()
```

```python
for i in num_col:
  sns.boxplot(vg[i])
  plt.show()
```
